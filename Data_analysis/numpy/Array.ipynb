    ## 1> reshape() it convert the 1D array shape in to 2d or multi D

    ## 3> min() it return the  small/ minimum  value in the array
    ## 4> argmin() it return the index of the minimum value in the array
    ## 5> argmax() it return the index of the maximum value in the array
    ## 6> sum()  it return the sum of the array elements in the array

## axis =1 it used to get minimum value from each and every row in the array
## axis =0 it used to get minimum value from each and every column in the array

# axis =1 - minimum  value from all row
# axis =0  minimum value from all column

```python
a1 = np.array([
    [1,21,2,45,22,56,5],
    [89,34,63,23,14,6,0]   # added 0
])
a1
```

```python
a1.min(axis=1)
```

```python
a1.min(axis=0)
```
