```python
import pandas as pd
import numpy as np
```

# Series : Collection of labelled data

```python
s1 = pd.Series([1,2,3,4,5])
s1
```

```python
s2 = pd.Series([10,20,30,40],index = ['a1','a2','a3','a4'])
s2
```

```python
topper = pd.Series({"IT":78 , "MECH":90,"Civil":90})
topper
```

## Operation on series

```python
s1 = pd.Series([10,20,30,40])
s2 = pd.Series([1,2,3,4])
s1+s2
```

## Labelledwise Operations


```python
s1 = pd.Series([10,20,30,40] ,index=["A","B","C","D"])
s2 = pd.Series([1,2,3,4],index=["C","D","G","H"])
s1+s2
```
